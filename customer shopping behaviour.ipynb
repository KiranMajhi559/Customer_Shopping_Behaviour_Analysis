```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower() 
df.columns = df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
df = df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.columns
```

```python
# creating age group column
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group']=pd.qcut(df['age'],q = 4,labels = labels) # qcut splits the ages into 4 equal sized groups based on the data distribution 
```

```python
df[['age','age_group']].head(10)
```

```python
df[['frequency_of_purchases']].head(10) # [[]] it will give with the column name
```

```python
# creating purchase_frequency_days coulmns 

frequency_mapping = {
    'Fortnightly':14,
    'Weekly':7,
    'Monthly':30,
    'Quarterly':90,
    'Bi-Weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping) # map will reapalce the texts in the columns
```

```python
df[['purchase_frequency_days','frequency_of_purchases']].head(15) 
```

```python
# check both the columns are same or not with the values
(df['discount_applied']==df['promo_code_used']).all()
```

```python
# remove one column
df = df.drop('promo_code_used',axis=1)
```
